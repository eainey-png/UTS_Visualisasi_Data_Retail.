# Analisis Awal Dataset Supermarket Sales

Notebook untuk pemeriksaan data, pembersihan sederhana, dan eksplorasi sebelum dashboard. Pastikan `dataset/supermarket_sales.csv` sudah tersedia.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

DATA_PATH = Path('../dataset/supermarket_sales.csv')
df = pd.read_csv(DATA_PATH)
df.head()
```

## 1. Memahami struktur data

```python
print('Ukuran data:', df.shape)
display(df.info())
display(df.describe(include='all').T)
```

## 2. Memeriksa kualitas data

```python
print('Duplikat:', df.duplicated().sum())
display(df.isna().sum().to_frame('Jumlah nilai kosong'))
```

## 3. Persiapan tipe data

```python
df.columns = [c.strip() for c in df.columns]
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
for col in ['Unit price','Quantity','Tax 5%','Total','cogs','gross income','Rating']:
    df[col] = pd.to_numeric(df[col], errors='coerce')
df = df.drop_duplicates().dropna(subset=['Date','Total','Quantity'])
print('Ukuran setelah pembersihan:', df.shape)
```

## 4. Ringkasan berdasarkan kategori

```python
display(df.groupby('Product line')['Total'].agg(['count','sum','mean']).sort_values('sum', ascending=False))
```

## 5. Ringkasan berdasarkan kota

```python
display(df.groupby('City')['Total'].agg(['count','sum','mean']).sort_values('sum', ascending=False))
```

## 6. Tren bulanan

```python
monthly = df.assign(Month=df['Date'].dt.to_period('M').astype(str)).groupby('Month')['Total'].sum()
monthly.plot(kind='line', marker='o', title='Total Pendapatan per Bulan', figsize=(8,4))
plt.ylabel('Total pendapatan')
plt.grid(True, alpha=.3)
plt.show()
```
